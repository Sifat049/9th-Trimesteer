```python
sample_input = """6
S 6 0
A 6 0
B 1 0
C 2 0
D 1 0
G 0 0
9
S A 1
S C 2
S D 4
A B 2
B A 2
B G 1
C S 2
C G 4
D G 4
S
G
"""

with open("/content/input.txt", "w") as f:
    f.write(sample_input)

print("input.txt created successfully!")
```

```python
import math
from queue import PriorityQueue



# Read Input


coords = {}
adjlist = {}


with open("input.txt", "r") as f:


    V = int(f.readline())

    for i in range(V):

        strs = f.readline().split()

        nid = strs[0]
        x = int(strs[1])
        y = int(strs[2])

        coords[nid] = (x, y)
        adjlist[nid] = []

    # edges
    E = int(f.readline())


    for i in range(E):

        strs = f.readline().split()

        n1 = strs[0]
        n2 = strs[1]
        cost = int(strs[2])

        adjlist[n1].append((n2, cost))

    # Start and goal
    startnid = f.readline().strip()
    goalnid = f.readline().strip()



# Heuristic Function


def heuristic(nid):

    x1, y1 = coords[nid]
    x2, y2 = coords[goalnid]

    # Euclidean distance
    return math.sqrt(
        (x2 - x1) ** 2 +
        (y2 - y1) ** 2
    )


# State Class


class State:

    def __init__(self, nid, parent, g, f):

        self.nid = nid
        self.parent = parent
        self.g = g
        self.f = f

    # PriorityQueue uses f value
    def __lt__(self, other):

        return self.f < other.f


# A* Search


def astar():

    minQ = PriorityQueue()

    # Initial state
    g = 0
    h = heuristic(startnid)
    f = g + h

    startState = State(
        startnid,
        None,
        g,
        f
    )

    minQ.put(startState)

    # A* search
    while not minQ.empty():

        # Get state with minimum f
        currstate = minQ.get()

        # Goal test
        if currstate.nid == goalnid:

            path = []

            cost = currstate.g

            # Backtrack using parent
            while currstate is not None:

                path.append(currstate.nid)
                currstate = currstate.parent

            # Reverse path
            path.reverse()

            print("Solution path:", " - ".join(path))
            print("Solution cost:", cost)

            return

        # Expand adjacent nodes
        for M, edge_cost in adjlist[currstate.nid]:

            # Actual cost
            g = currstate.g + edge_cost


            h = heuristic(M)

            # Total estimated cost
            f = g + h

            #  new state
            newstate = State(
                M,
                currstate,
                g,
                f
            )

            # Add to priority queue
            minQ.put(newstate)


    print("No Solution Found")



astar()
```
