```python
import math
import csv
import time
from collections import defaultdict
from heapq import heappush, heappop


COORDINATES_FILE = "/content/Coordinates.csv"
DISTANCE_FILE = "/content/distances.csv"

# Read Coordinates


coordinates = {}

with open(COORDINATES_FILE, "r") as file:
    reader = csv.reader(file)

    next(reader)  # Skip header

    for star_name, x, y, z in reader:
        coordinates[star_name] = (
            int(x),
            int(y),
            int(z)
        )

# Create Adjacency List


adjacency_list = defaultdict(list)

with open(DISTANCE_FILE, "r") as file:
    reader = csv.reader(file)

    for source, destination, dist in reader:
        adjacency_list[source].append(
            (destination, int(dist))
        )

# Euclidean Heuristic


def heuristic(star, destination):

    x1, y1, z1 = coordinates[star]
    x2, y2, z2 = coordinates[destination]

    return math.sqrt(
        (x1 - x2) ** 2 +
        (y1 - y2) ** 2 +
        (z1 - z2) ** 2
    )


# Dijkstra

def dijkstra(source, destination):

    start_time = time.perf_counter()

    priority_queue = [(0, source)]

    best_distance = {
        source: 0
    }

    parent = {}

    visited = set()

    nodes_expanded = 0

    while priority_queue:

        dist, current = heappop(priority_queue)

        if current in visited:
            continue

        visited.add(current)
        nodes_expanded += 1

        # Goal reached
        if current == destination:

            path = []

            node = current

            while node != source:
                path.append(node)
                node = parent[node]

            path.append(source)

            path.reverse()

            end_time = time.perf_counter()

            return {
                "path": path,
                "distance": dist,
                "time": end_time - start_time,
                "expanded": nodes_expanded
            }

        # Explore neighbors
        for neighbor, edge_cost in adjacency_list[current]:

            new_distance = dist + edge_cost

            if new_distance < best_distance.get(
                neighbor,
                float("inf")
            ):

                best_distance[neighbor] = new_distance

                parent[neighbor] = current

                heappush(
                    priority_queue,
                    (new_distance, neighbor)
                )

    end_time = time.perf_counter()

    return {
        "path": None,
        "distance": None,
        "time": end_time - start_time,
        "expanded": nodes_expanded
    }


# A*


def a_star(source, destination):

    start_time = time.perf_counter()

    # f = g + h
    h = heuristic(source, destination)

    priority_queue = [
        (h, 0, source)
    ]

    best_g = {
        source: 0
    }

    parent = {}

    visited = set()

    nodes_expanded = 0

    while priority_queue:

        f, g, current = heappop(priority_queue)

        if current in visited:
            continue

        visited.add(current)
        nodes_expanded += 1

        # Goal reached
        if current == destination:

            path = []

            node = current

            while node != source:
                path.append(node)
                node = parent[node]

            path.append(source)

            path.reverse()

            end_time = time.perf_counter()

            return {
                "path": path,
                "distance": g,
                "time": end_time - start_time,
                "expanded": nodes_expanded
            }

        # Explore neighbors
        for neighbor, edge_cost in adjacency_list[current]:

            new_g = g + edge_cost

            if new_g < best_g.get(
                neighbor,
                float("inf")
            ):

                best_g[neighbor] = new_g

                parent[neighbor] = current

                h = heuristic(
                    neighbor,
                    destination
                )

                f = new_g + h

                heappush(
                    priority_queue,
                    (f, new_g, neighbor)
                )

    end_time = time.perf_counter()

    return {
        "path": None,
        "distance": None,
        "time": end_time - start_time,
        "expanded": nodes_expanded
    }


# Print Result


def print_result(name, result):

    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)

    if result["path"] is None:

        print("Destination not reachable.")

    else:

        print(
            "Path:",
            " -> ".join(result["path"])
        )

        print(
            "Total Distance:",
            result["distance"]
        )

    print(
        "Nodes Expanded:",
        result["expanded"]
    )

    print(
        "Execution Time:",
        result["time"]
    )


# Test Cases


test_cases = [

    ("Sun", "Upsilon Andromedae"),

    ("Sun", "61 Virginis"),

    ("TRAPPIST-1", "55 Cancri")
]


for source, destination in test_cases:

    print("\n\n")
    print("#" * 60)
    print("SOURCE:", source)
    print("DESTINATION:", destination)
    print("#" * 60)

    dijkstra_result = dijkstra(
        source,
        destination
    )

    astar_result = a_star(
        source,
        destination
    )

    print_result(
        "DIJKSTRA",
        dijkstra_result
    )

    print_result(
        "A*",
        astar_result
    )
```
